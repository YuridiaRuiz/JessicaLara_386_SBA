```python
# 6. Print  the "total_profit" and “month_number” columns only.
subset_df = df[['total_profit', 'month_number']]
print(subset_df)
```

##### Read the total profit of all months and show it using the Bar plot. 
- The generated Bar plot must include the following properties: 
- X label name = Month number.
- Y label name = Total profit.
- Title = Company profit per month.


```python
# 7. Create a bar chart to visualize the total profit for each month.
plt.figure(figsize=(10, 6))
plt.bar(subset_df['month_number'], subset_df['total_profit'])
plt.title('Company profit per month', fontsize=16, fontweight='bold')
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.xticks(subset_df['month_number'])  # Set x-ticks to month numbers
plt.yticks(np.arange(0, max(subset_df['total_profit']) + 100000, 100000))  # Set y-ticks with a step of 100000
plt.grid(alpha=0.3, linestyle='--')  
# Show the plot
plt.show()
```

##### Read the total profit of all of the months, and show the line plot with the following style properties:
- The generated line plot must include the following Style properties: 
- The line style should be dotted and the line color should be red.
- Show the legend at the lower-right location.
- X label name = Month number
- Y label name = Total profit.
- Title = Company profit per month.
- Add a circle marker; the line marker color = black. 
- The line width should be 3.


```python
# 8. Create a line plot to visualize the relationship between month number and total profit.
plt.figure(figsize=(10, 6))
plt.plot(subset_df['month_number'], subset_df['total_profit'], linestyle='dotted', color='red', label= 'Profit data of last year',marker='o', mfc='black', mec='red', linewidth=3)
plt.title('Company profit per month', fontsize=16, fontweight='bold')
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.legend(loc='lower right') #show legend at the lower-right corner
plt.xticks(subset_df['month_number'])  # Set x-ticks to month numbers
plt.yticks(np.arange(0, max(subset_df['total_profit']) + 100000, 100000))  # Set y-ticks with a step of 100000
plt.grid(alpha=0.3, linestyle='--')
# Show the plot
plt.show()
```

##### Print all of the product sales data and show it using a multi-line plot.
- Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).



```python
# 9.Create a line chart to visualize montly unid sold for each product.
product_columns = ['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer']
plt.figure(figsize=(10, 6))

for product in product_columns:
    plt.plot(df['month_number'], df[product], marker='o', label=product.capitalize()+' Sales Data')

plt.xlabel('Month number')
plt.ylabel('Sales units in number')
plt.title('Sales Data', fontsize=16, fontweight='bold')
plt.xticks(df['month_number'])
plt.legend()
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```

##### Read “bathingsoap” sales data for each month and show it using a scatter plot.
- Add a grid to the plot. The grid line style should be “–“.



```python
# Create a scatter plot of bathinsoap sales data
plt.figure(figsize=(10, 6))
plt.scatter(df['month_number'], df['bathingsoap'], color='blue', label='Bathing soap Sales data', marker='o')
plt.title('Bathing Soap Sales Data', fontsize=16, fontweight='bold')
plt.xlabel('Month number')
plt.ylabel('Number of units sold')
plt.xticks(df['month_number'])
plt.yticks(np.arange(0, max(df['bathingsoap']) + 2000, 2000))  # Set y-ticks with a step of 2000
plt.legend(loc='upper left')
plt.grid(alpha=0.3, linestyle='-')
plt.show()
```

#### 1. Given Dataset, create a line chart
- A line chart of date versus temperature must include the following properties: 
- X label name = Date
- Y label name = temperature
- Title = Date-wise Temperature


```python
# Dataset for the line chart
date=["25/12","26/12","27/12"]
temp=[8.5,10.5,6.8]
```

```python
# Create a line chart to visualize the temperature data over the given dates.
plt.figure(figsize=(6, 4))
plt.plot(date, temp, marker='o', color='orange', label='Temperature Data',mfc='red', mec='orange', linewidth=1, linestyle='--')
plt.xlabel('Date')
plt.ylabel('temperature')
plt.title('Date-wise Temperature', fontsize=16, fontweight='bold')
plt.grid(linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.show()
```

##### 2. Given a dataset, create a chart 
- Show the average weight against the average height (use a line chart or any other chart):
- The generated line plot must include the following Style properties:
- The line style should be dash-dot and the line color should be green.
- Show the legend at the lower-right location.
- X label name = Weight in kg.
- Y label name = Height in cm.
- Title = Average weight with respect to average height.
- Add a circle marker: marker size = 10; circle marker color = green.


```python
# Given Dataset:
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]


plt.figure(figsize=(10, 6))
plt.plot(weight, height, linestyle='-.', marker='o', mfc='green', mec='green', ms=10, color='green', linewidth=2, label='Average weight with respect to average height')
plt.xlabel('Weight in Kg')
plt.ylabel('Height in cm')
plt.title('Average weight with respect to average height')
plt.grid(alpha=0.3, linestyle='--')
plt.legend(loc='lower right')
plt.show()
```
